```python
import numpy as np
import json
import matplotlib.pyplot as plt
```

```python
f = open('data/models/novas_av_re_no_vis/v1/test_stats_test-unseen.json')  
data = json.load(f)
for i in data.keys():
    print(i)
f.close()

f = open('data/models/novas_av_re_no_vis/v1/test_stats_test-unseen.json')  
data_no_vis = json.load(f)
for i in data_no_vis.keys():
    print(i)
f.close()
```

```python
pred = np.array(data['pred_rt60'])
tgt = np.array(data['tgt_rt60'])

pred_no_vis = np.array(data_no_vis['pred_rt60'])
tgt_no_vis = np.array(data_no_vis['tgt_rt60'])
```

```python
fig, axs = plt.subplots(3, 2)
fig.set_size_inches(15, 16)

axs[0, 0].hist(pred, bins=100)
axs[0, 0].set_title("Pred RT60 Distribution (vis)")
axs[0, 0].set_xlabel("RT60")
axs[0, 0].set_ylabel("Frequency")

axs[1, 0].hist(tgt, bins=100)
axs[1, 0].set_title("Tgt RT60 Distribution (vis)")
axs[1, 0].set_xlabel("RT60")
axs[1, 0].set_ylabel("Frequency")

axs[2, 0].hist(tgt-pred, bins=100)
axs[2, 0].set_title("Paired RT60 Difference Distribution (vis)")
axs[2, 0].set_xlabel("RT60")
axs[2, 0].set_ylabel("Frequency")

axs[0, 1].hist(pred_no_vis, bins=100)
axs[0, 1].set_title("Pred RT60 Distribution (no vis)")
axs[0, 1].set_xlabel("RT60")
axs[0, 1].set_ylabel("Frequency")

axs[1, 1].hist(tgt_no_vis, bins=100)
axs[1, 1].set_title("Tgt RT60 Distribution (no vis)")
axs[1, 1].set_xlabel("RT60")
axs[1, 1].set_ylabel("Frequency")

axs[2, 1].hist(tgt_no_vis-pred_no_vis, bins=100)
axs[2, 1].set_title("Paired RT60 Difference Distribution (no vis)")
axs[2, 1].set_xlabel("RT60")
axs[2, 1].set_ylabel("Frequency")
```
